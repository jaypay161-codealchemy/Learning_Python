````python
# # Instructions
# 1. create a list called "tasks" and populate it with initial tasks: "Planning","Design","Coding","Testing".
# 2. create another list called "time_spent" to store the time spent on each task. Intialize it with zeros.
# 3. create function called "log_time" that takes a task name and an amount of time. It will find the task in the "task" list and update the corresponding time in "time_spent".
# 4. Use a "while" loop to repeatedly ask the user for a task and time spent on that task until they choose to exit.
# 5. Use a "for" loop to go through the "tasks" and "time_spent" lists, printing out a summary of time spent on each task.

tasks = ["Planning","Design","Coding", "Testing"]
time_spent = [0, 0, 0, 0]

def log_time(task, time):
  for t in range(len(tasks)):
    if tasks[t] == task:
      time_spent[t] += time



while True:
  print("Tasks: ", tasks)
  task = input("Enter task to log time for it / or enter exit to stop the program.")

  if task == "exit":
     break
  else:
    if task in tasks:
      time = float(input("Enter time spent on task: "))
      log_time(task, time)
    else:
      print("Oops Task not found, please enter a valid task")

for i in range(len(tasks)):
  print(f"Time spent on {tasks[i]} is {time_spent[i]}")


````

````python
from typing import Any
# Step 1 : Initial Menu
menu = {
    "Burger":("Main", 10.5),
    "Soup":("Appetizer", 5.0),
    "Ice cream":("Dessert", 3.5),
    "Salad": ("Appetizer", 6.0),
}

# Step 2 : Add dishes
menu['steak']=["Main", 20.0]
menu['Soda']=["Main", 2.0]


# Step 3 : Remove dishes
menu.pop("Salad")
#print(menu.items())

# Step 4 : Display menu function
def display_menu(menu):
    print("Menu:")
    for D, C in menu.items():
        print(f"{D}: {C[0]} - ${C[1]} \n")

# display_menu(menu)


# Step 5 : Count dish type function

def count_dish_type(menu):
  dish_per_category = {}
  for dish, info in menu.items():
    if info[0] in dish_per_category:
      dish_per_category[info[0]] += 1
    else:
      dish_per_category[info[0]] = 1

  return dish_per_category

# print(count_dish_type(menu))


# Step 6 : Update Price function
display_menu(menu)

def update_price(menu, dish, new_price):
  if dish in menu:
    menu[dish][1] = new_price
    print(f"\n \n Price of {dish} has been updated to ${new_price}")
  else:
    print("This dish does not exist")

update_price(menu, "steak", 25.0)
display_menu(menu)

````

````python
# step 0: sample chat logs
chat_logs = [
    "My order is delayed",
    "I want to return my purchase",
    "The app is crashing frequently",
    "Payment issues, Please Help",
    "Need help with account recovery",
    "Delivery was incomplete",
    "Can't login to my account",
    "Having trouble with checkout"
]

# step 1: Function to categorized the issues

categories = {"Order":0, "App":0, "Payment":0, "Account":0, "Delivery":0 }
def categorized_issues(chat_logs):
    for log in chat_logs:
      log_lower = log.lower()
      for category in categories.keys():
        if category.lower() in log_lower:
          categories[category] += 1


# step 2: Function to search by keyword
def keyword_search(keyword, logs):
  keyword = keyword.lower()
  results =[]
  for log in logs:
    log_lower = log.lower()
    if keyword in log_lower:
      results.append(log)


  return results



# step 3: Display and search
print(categories)
categorized_issues(chat_logs)
print(categories)

keyword = input("Enter a keyword to search for: ")
results = keyword_search(keyword, chat_logs)
print(results)
````

**Inheritance of classes**

````python
class vehicle():
  def start_engine(self,name,model,year,cost):
    print("Engine started !",name,model,year,cost)

class car(vehicle):
  pass

x = car()
x.start_engine("Maruti","Model:"+str("VXL"),"Year:"+str(2015),"Rs.4,00,000/- incl tax")
````

````python
class BankAccount():
  def __init__(self,balance):
    self.__balance = balance

  def deposit(self,amount):
    self.__balance += amount

  def get_balance(self):
    return self.__balance

x1 = BankAccount(1000)
x1.deposit(500)
print(x1.get_balance())
#print(x1.__balance)...."Encapsulation concept"

````

````python
# Task 1: Using Python Loop
# Create a list called

import numpy as np

monthly_sales = np.random.randint(1000, 2000, 12)

print("List of monthly sales :",monthly_sales)

# coding done in normal python way
discount_sales = []

for e in monthly_sales:
  discount_sales.append(e*0.9) # 10% discount

print("\n discount values using normal python",discount_sales)


# coding done in numpy vectorization

discount_sales_np = monthly_sales*0.9

print("\n discount values using numpy vectorization",discount_sales_np)
````
