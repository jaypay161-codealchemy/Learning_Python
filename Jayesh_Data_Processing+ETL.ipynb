## Data cleaning Part 2

```python
import pandas as pd
from scipy import stats

df = pd.DataFrame({
    "Transaction_Amount": [100, 150, 200, 250, 300, 9999]
})
```

```python
df
```

```python
df['Z_Score'] = stats.zscore(df['Transaction_Amount'])
```

```python
df
```

```python
df['Winsorized_Amount'] = stats.mstats.winsorize(df['Transaction_Amount'], limits=[0.05, 0.05])
df
```

```python
threshold = 500

df_truncate = df[df['Transaction_Amount'] <= threshold]
df_truncate
```

Feature ENGINEERING

```python
df1 = pd.DataFrame({
    "age": [25,30,35,40,45,75]
})

df1
```

```python
bins = [0, 30, 50, 100]
labels = ['Young', 'Middle-aged', 'Senior']

df1['bin_age'] = pd.cut(df1['age'], bins=bins, labels=labels)
df1
```

```python
from sklearn.preprocessing import PolynomialFeatures

df2 = pd.DataFrame({"size": [550,700,900], "rooms": [2,3,4]})

poly_f = PolynomialFeatures(degree=2, include_bias=False)

df_poly = pd.DataFrame(poly_f.fit_transform(df2), columns=poly_f.get_feature_names_out())
df_poly
```
